# filling missing value with regression imputation

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.experimental import enable_hist_gradient_boosting  
from sklearn.ensemble import HistGradientBoostingRegressor     # i use this to handle missing values in X features
from sklearn.model_selection import KFold,cross_val_score, train_test_split
from sklearn.metrics import mean_squared_error, make_scorer
```

```python
# splitting data into missing and non missing values
# splitting the data into test and train (train with data without missing value and test with data with missing values)

def regression_imputation(df, target_col):
    missing_value = df[target_col].isna()     #rows where we have the missing values
    if not missing_value.any():
        return df

    train_set = df[~missing_value]      # The ~ operator is the bitwise NOT operator, which inverts the boolean values   
    test_set = df[missing_value]

    X_train = train_set.drop(columns=[target_col])
    y_train = train_set[target_col]
    X_test = test_set.drop(columns=[target_col])

    model = HistGradientBoostingRegressor()
    model.fit(X_train, y_train)
    predicted_values = model.predict(X_test)
    
    df.loc[missing_value, target_col] = predicted_values  # update the df with the predicted values
    
    return df

df = regression_imputation(df, 'loc')
df = regression_imputation(df, 'title')
df = regression_imputation(df, 'bedroom')
df = regression_imputation(df, 'bathroom')
df = regression_imputation(df, 'parking_space')
```

```python
# dealing with the skewed price (log transformation)

import numpy as np
df['log_price']=np.log1p(df['price'])
```

```python
X=df[['loc','title','bedroom','bathroom','parking_space']]
y=df['log_price']
```

```python
# train test split
X_train, X_test, y_train, y_test= train_test_split(X, y, test_size=.20, random_state=42,shuffle=True)
```

```python
model= HistGradientBoostingRegressor()
```

```python
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
mse = mean_squared_error(y_test, y_pred)
mse
```

```python
# convert the log price back to its original price

y_pred_original = np.expm1(y_pred)
y_test_original = np.expm1(y_test)
```

```python
mse_original_price = round(mean_squared_error(y_test_original, y_pred_original),3)
mse_original_price
```

```python
# using the original price with the KFold cross validation

X1=df[['loc','title','bedroom','bathroom','parking_space']]
y1=df['price']
```

```python
kf=KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
mse = make_scorer(mean_squared_error)
```

```python
scores = cross_val_score(model, X1, y1, cv=kf, scoring=mse)
```

```python
print(f"MSE Score for each fold: {[round(score, 3) for score in scores]}")
```
